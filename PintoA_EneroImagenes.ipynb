```python
#########################
#        LIBRERÍAS      #
#########################
import os
import time
import copy
import random
import numpy as np
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
import torchvision
import torchvision.transforms as transforms
from torchvision import datasets, models
from tqdm import tqdm

# Montar Google Drive
from google.colab import drive
drive.mount('/content/drive')

#########################
#     CONFIGURACIÓN     #
#########################
# Directorios
base_dir = '/content/drive/MyDrive/IA_ProyectoAgosto/datasetAffectNet/AffectNet'
train_dir = os.path.join(base_dir, 'train')
val_dir   = os.path.join(base_dir, 'val')
test_dir  = os.path.join(base_dir, 'test')

# Definir la carpeta para guardar logs/estadísticas
logs_dir = '/content/drive/MyDrive/IA_ProyectoAgosto/LogsEnero/Affectnet_logs'
if not os.path.exists(logs_dir):
    os.makedirs(logs_dir)

# Hiperparámetros
num_classes = 6   # (angry=0, disgust=1, fear=2, happy=3, sad=4, surprise=5)
batch_size  = 32
num_workers = 2
epochs      = 5   # Ajustar la cantidad de épocas según sea necesario
learning_rate = 1e-4
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
print("Dispositivo usado:", device)

#########################
#     TRANSFORMACIONES  #
#########################
# Normalmente se usan estadísticas de ImageNet si se usan modelos preentrenados
mean = [0.485, 0.456, 0.406]
std  = [0.229, 0.224, 0.225]

# Definir transformaciones de entrenamiento y validación/test
data_transforms = {
    'train': transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.RandomHorizontalFlip(),
        transforms.ToTensor(),
        transforms.Normalize(mean, std)
    ]),
    'val': transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(mean, std)
    ]),
    'test': transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(mean, std)
    ])
}

#########################
#       DATASETS        #
#########################
image_datasets = {
    'train': datasets.ImageFolder(train_dir, transform=data_transforms['train']),
    'val':   datasets.ImageFolder(val_dir,   transform=data_transforms['val']),
    'test':  datasets.ImageFolder(test_dir,  transform=data_transforms['test'])
}

# DataLoaders
dataloaders = {
    'train': DataLoader(image_datasets['train'], batch_size=batch_size, shuffle=True,  num_workers=num_workers),
    'val':   DataLoader(image_datasets['val'],   batch_size=batch_size, shuffle=False, num_workers=num_workers),
    'test':  DataLoader(image_datasets['test'],  batch_size=batch_size, shuffle=False, num_workers=num_workers)
}

dataset_sizes = {
    x: len(image_datasets[x]) for x in ['train', 'val', 'test']
}

class_names = image_datasets['train'].classes
print("Clases encontradas en train:", class_names)



for split in ['train', 'val', 'test']:
    print(f"=== Conteo de imágenes en {split} ===")
    dataset = image_datasets[split]
    for class_name in dataset.classes:
        # 'dataset.class_to_idx' mapea el nombre de la clase a un índice entero
        class_idx = dataset.class_to_idx[class_name]

        # Contar cuántas muestras tienen la etiqueta 'class_idx'
        count = sum(1 for _, label in dataset.samples if label == class_idx)

        print(f"Clase: {class_name:<15} -> {count} imágenes")
    print()





#########################
# FUNC. AUXILIARES      #
#########################
def train_model(model, criterion, optimizer, dataloaders, device, num_epochs=5, logs_filename='training_log.txt'):

    import time
    import copy
    import os
    """
    Entrena el modelo y guarda logs de pérdida y exactitud por época.
    """
    since = time.time()

    best_model_wts = copy.deepcopy(model.state_dict())
    best_acc = 0.0

    train_loss_history = []
    val_loss_history   = []
    train_acc_history  = []
    val_acc_history    = []

    for epoch in range(num_epochs):
        print(f'Epoch {epoch+1}/{num_epochs}')
        print('-' * 10)

        # Cada época tiene fase de train y val
        for phase in ['train', 'val']:
            if phase == 'train':
                model.train()
            else:
                model.eval()

            running_loss = 0.0
            running_corrects = 0
            # Aquí usamos tqdm para que nos muestre progreso en cada batch.
            # 'unit="batch"' indica que la barra dirá "xx% [xx/<total>]" por lotes.
            with tqdm(dataloaders[phase], desc=f"{phase} - epoch {epoch+1}", unit="batch") as tepoch:
                # Iterar por los datos
                for inputs, labels in tepoch:
                    inputs = inputs.to(device)
                    labels = labels.to(device)

                    # Poner gradientes a cero
                    optimizer.zero_grad()

                    # forward
                    # track history solo en entrenamiento
                    with torch.set_grad_enabled(phase == 'train'):
                        outputs = model(inputs)
                        _, preds = torch.max(outputs, 1)
                        loss = criterion(outputs, labels)

                        # backward + optimizar solo en entrenamiento
                        if phase == 'train':
                            loss.backward()
                            optimizer.step()

                    # Estadísticas
                    running_loss += loss.item() * inputs.size(0)
                    running_corrects += torch.sum(preds == labels.data)

                    epoch_loss = running_loss / len(dataloaders[phase].dataset)
                    epoch_acc = running_corrects.double() / len(dataloaders[phase].dataset)


            print(f'{phase} Loss: {epoch_loss:.4f} Acc: {epoch_acc:.4f}')

            # Guardar los resultados para graficar
            if phase == 'train':
                train_loss_history.append(epoch_loss)
                train_acc_history.append(epoch_acc.item())
            else:
                val_loss_history.append(epoch_loss)
                val_acc_history.append(epoch_acc.item())

            # Realizar deep copy del mejor modelo
            if phase == 'val' and epoch_acc > best_acc:
                best_acc = epoch_acc
                best_model_wts = copy.deepcopy(model.state_dict())

        print()

        # Guardar logs en archivo de texto
        with open(os.path.join(logs_dir, logs_filename), 'a') as f:
            f.write(f"Epoch {epoch+1}/{num_epochs}, "
                    f"Train Loss: {train_loss_history[-1]:.4f}, "
                    f"Train Acc: {train_acc_history[-1]:.4f}, "
                    f"Test Loss: {val_loss_history[-1]:.4f}, "
                    f"Test Acc: {val_acc_history[-1]:.4f}\n")

    time_elapsed = time.time() - since
    print(f'Entrenamiento completo en {time_elapsed // 60:.0f}m {time_elapsed % 60:.0f}s')
    print(f'Mejor Accruracy en Val: {best_acc:4f}')

    # Cargar los mejores pesos al modelo
    model.load_state_dict(best_model_wts)

    # Retor para evaluación y para poder graficar la historia
    return model, (train_loss_history, val_loss_history, train_acc_history, val_acc_history)

def visualize_metrics(history, title_prefix="Model"):
    """
    Grafica las métricas de Loss y Accuracy a lo largo de las épocas.
    """
    train_loss, val_loss, train_acc, val_acc = history

    epochs = range(1, len(train_loss) + 1)

    plt.figure(figsize=(14, 5))

    # Loss
    plt.subplot(1, 2, 1)
    plt.plot(epochs, train_loss, 'r-', label='Train Loss')
    plt.plot(epochs, val_loss, 'b-', label='Val Loss')
    plt.title(f'{title_prefix} - Loss vs Épocas')
    plt.xlabel('Épocas')
    plt.ylabel('Loss')
    plt.legend()

    # Accuracy
    plt.subplot(1, 2, 2)
    plt.plot(epochs, train_acc, 'r-', label='Train Accuracy')
    plt.plot(epochs, val_acc, 'b-', label='Val Accuracy')
    plt.title(f'{title_prefix} - Accuracy vs Épocas')
    plt.xlabel('Épocas')
    plt.ylabel('Accuracy')
    plt.legend()

    plt.show()

def visualize_predictions(model, dataloader, device, num_images=6):
    """
    Muestra imágenes del dataset junto a su predicción.
    """
    model.eval()
    images_so_far = 0
    fig = plt.figure(figsize=(10, 10))

    with torch.no_grad():
        for i, (inputs, labels) in enumerate(dataloader):
            inputs = inputs.to(device)
            labels = labels.to(device)

            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)

            for j in range(inputs.size(0)):
                images_so_far += 1
                ax = plt.subplot(num_images//3, 3, images_so_far)
                ax.axis('off')
                ax.set_title(f'Real: {class_names[labels[j]]}\nPred: {class_names[preds[j]]}')
                inp = inputs.cpu().data[j]
                inp = inp.numpy().transpose((1, 2, 0))
                inp = std * inp + mean  # desnormalizar
                inp = np.clip(inp, 0, 1)
                plt.imshow(inp)

                if images_so_far == num_images:
                    return


#########################
#     MODELOS (2)       #
#########################
# 1. ResNet18 preentrenado
model_resnet = models.resnet18(weights='IMAGENET1K_V1')  # PyTorch 2.x
# Ajustar capa final al número de clases
num_ftrs = model_resnet.fc.in_features
model_resnet.fc = nn.Linear(num_ftrs, num_classes)
model_resnet = model_resnet.to(device)

# Definir loss y optimizador
criterion = nn.CrossEntropyLoss()
optimizer_resnet = optim.Adam(model_resnet.parameters(), lr=learning_rate)

# Entrenar el modelo 1
trained_resnet, history_resnet = train_model(
    model_resnet,
    criterion,
    optimizer_resnet,
    {'train': dataloaders['train'], 'val': dataloaders['val']},
    device,
    num_epochs=epochs,
    logs_filename='resnet_training_log.txt'
)

# Graficar sus métricas
visualize_metrics(history_resnet, title_prefix="ResNet18")

# 2. VGG16 preentrenado
model_vgg = models.vgg16(weights='IMAGENET1K_V1')
# Ajustar la última capa
num_ftrs = model_vgg.classifier[6].in_features
model_vgg.classifier[6] = nn.Linear(num_ftrs, num_classes)
model_vgg = model_vgg.to(device)

optimizer_vgg = optim.Adam(model_vgg.parameters(), lr=learning_rate)

trained_vgg, history_vgg = train_model(
    model_vgg,
    criterion,
    optimizer_vgg,
    {'train': dataloaders['train'], 'val': dataloaders['val']},
    device,
    num_epochs=epochs,
    logs_filename='vgg_training_log.txt'
)

visualize_metrics(history_vgg, title_prefix="VGG16")

#########################
# EVALUACIÓN EN TEST  #
#########################
def test_model(model, dataloader, device):
    model.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs = inputs.to(device)
            labels = labels.to(device)
            outputs = model(inputs)
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
    accuracy = 100 * correct / total
    return accuracy

resnet_test_accuracy = test_model(trained_resnet, dataloaders['test'], device)
vgg_test_accuracy    = test_model(trained_vgg, dataloaders['test'], device)

print("Test Accuracy - ResNet18:", resnet_test_accuracy)
print("Test Accuracy - VGG16:   ", vgg_test_accuracy)

#########################
#  EJEMPLOS DE PREDIC.  #
#########################
# Visualizar algunas predicciones reales vs. predichas en test
visualize_predictions(trained_resnet, dataloaders['test'], device, num_images=6)
visualize_predictions(trained_vgg, dataloaders['test'], device, num_images=6)
```

```python
#####################################
#        IMPORTAR LIBRERÍAS         #
#####################################
import os
import time
import copy
import random
import numpy as np
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
import torchvision
import torchvision.transforms as transforms
from torchvision import datasets, models

# Barra de progreso
!pip install tqdm
from tqdm import tqdm

# Montar Google Drive
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

#####################################
#  CONFIGURACIÓN DE RUTAS Y LOGS    #
#####################################
base_dir = '/content/drive/MyDrive/IA_ProyectoAgosto/datasetAffectNet/AffectNet'  # Ajusta a tu ruta
train_dir = os.path.join(base_dir, 'train')
val_dir   = os.path.join(base_dir, 'val')
test_dir  = os.path.join(base_dir, 'test')

# Carpeta para guardar logs
logs_dir = '/content/drive/MyDrive/IA_ProyectoAgosto/LogsEnero/Affectnet_logs'
if not os.path.exists(logs_dir):
    os.makedirs(logs_dir)

#####################################
#       HIPERPARÁMETROS             #
#####################################
num_classes   = 6  # (angry=0, disgust=1, fear=2, happy=3, sad=4, surprise=5)
batch_size    = 32
num_workers   = 2
epochs        = 5
learning_rate = 1e-4

# Seleccionar dispositivo
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
print("Dispositivo usado:", device)

#####################################
#  TRANSFORMACIONES DE LAS IMÁGENES #
#####################################
# Estadísticas de normalización usuales de ImageNet
mean = [0.485, 0.456, 0.406]
std  = [0.229, 0.224, 0.225]

data_transforms = {
    'train': transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.RandomHorizontalFlip(),
        transforms.ToTensor(),
        transforms.Normalize(mean, std)
    ]),
    'val': transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(mean, std)
    ]),
    'test': transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(mean, std)
    ])
}

#####################################
#          PREPARAR DATASET         #
#####################################
image_datasets = {
    'train': datasets.ImageFolder(train_dir, transform=data_transforms['train']),
    'val':   datasets.ImageFolder(val_dir,   transform=data_transforms['val']),
    'test':  datasets.ImageFolder(test_dir,  transform=data_transforms['test'])
}

dataloaders = {
    'train': DataLoader(image_datasets['train'], batch_size=batch_size, shuffle=True,  num_workers=num_workers),
    'val':   DataLoader(image_datasets['val'],   batch_size=batch_size, shuffle=False, num_workers=num_workers),
    'test':  DataLoader(image_datasets['test'],  batch_size=batch_size, shuffle=False, num_workers=num_workers)
}

dataset_sizes = {x: len(image_datasets[x]) for x in ['train', 'val', 'test']}

class_names = image_datasets['train'].classes
print("Clases encontradas:", class_names)

################################################
#      FUNCIÓN DE ENTRENAMIENTO (con tqdm)     #
################################################
def train_model(model, criterion, optimizer, dataloaders, device, num_epochs=5, logs_filename='training_log.txt'):
    """
    Entrena el modelo y registra el mejor modelo basado en la exactitud de validación.
    """
    since = time.time()

    best_model_wts = copy.deepcopy(model.state_dict())
    best_acc = 0.0

    train_loss_history = []
    val_loss_history   = []
    train_acc_history  = []
    val_acc_history    = []

    for epoch in range(num_epochs):
        print(f'Epoch {epoch+1}/{num_epochs}')
        print('-' * 10)

        for phase in ['train', 'val']:
            if phase == 'train':
                model.train()
            else:
                model.eval()

            running_loss = 0.0
            running_corrects = 0

            # Barra de progreso por batches
            with tqdm(dataloaders[phase], desc=f"{phase} - epoch {epoch+1}", unit="batch") as tepoch:
                for inputs, labels in tepoch:
                    inputs = inputs.to(device)
                    labels = labels.to(device)

                    # Poner gradientes a cero
                    optimizer.zero_grad()

                    # Forward
                    with torch.set_grad_enabled(phase == 'train'):
                        outputs = model(inputs)
                        _, preds = torch.max(outputs, 1)
                        loss = criterion(outputs, labels)

                        # Backward + optimizar solo en entrenamiento
                        if phase == 'train':
                            loss.backward()
                            optimizer.step()

                    # Estadísticas
                    running_loss += loss.item() * inputs.size(0)
                    running_corrects += torch.sum(preds == labels.data)

            epoch_loss = running_loss / dataset_sizes[phase]
            epoch_acc = running_corrects.double() / dataset_sizes[phase]

            print(f'{phase} Loss: {epoch_loss:.4f} Acc: {epoch_acc:.4f}')

            if phase == 'train':
                train_loss_history.append(epoch_loss)
                train_acc_history.append(epoch_acc.item())
            else:
                val_loss_history.append(epoch_loss)
                val_acc_history.append(epoch_acc.item())

            # Guardar el mejor modelo según la métrica de validación
            if phase == 'val' and epoch_acc > best_acc:
                best_acc = epoch_acc
                best_model_wts = copy.deepcopy(model.state_dict())

        print()

        # Guardar logs en archivo
        with open(os.path.join(logs_dir, logs_filename), 'a') as f:
            f.write(f"Epoch {epoch+1}/{num_epochs}, "
                    f"Train Loss: {train_loss_history[-1]:.4f}, "
                    f"Train Acc: {train_acc_history[-1]:.4f}, "
                    f"Val Loss: {val_loss_history[-1]:.4f}, "
                    f"Val Acc: {val_acc_history[-1]:.4f}\n")

    time_elapsed = time.time() - since
    print(f'Entrenamiento completo en {time_elapsed // 60:.0f}m {time_elapsed % 60:.0f}s')
    print(f'Mejor Accuracy en Val: {best_acc:4f}')

    # Cargar los mejores pesos
    model.load_state_dict(best_model_wts)
    return model, (train_loss_history, val_loss_history, train_acc_history, val_acc_history)


######################################
#  FUNCIONES PARA GRAFICAR MÉTRICAS  #
######################################
def visualize_metrics(history, title_prefix="Model"):
    train_loss, val_loss, train_acc, val_acc = history
    epochs = range(1, len(train_loss) + 1)

    plt.figure(figsize=(14, 5))

    # Gráfica de pérdida
    plt.subplot(1, 2, 1)
    plt.plot(epochs, train_loss, 'r-', label='Train Loss')
    plt.plot(epochs, val_loss, 'b-', label='Test Loss')
    plt.title(f'{title_prefix} - Loss vs Épocas')
    plt.xlabel('Épocas')
    plt.ylabel('Loss')
    plt.legend()

    # Gráfica de exactitud
    plt.subplot(1, 2, 2)
    plt.plot(epochs, train_acc, 'r-', label='Train Accuracy')
    plt.plot(epochs, val_acc, 'b-', label='Test Accuracy')
    plt.title(f'{title_prefix} - Accuracy vs Épocas')
    plt.xlabel('Épocas')
    plt.ylabel('Accuracy')
    plt.legend()

    plt.show()

def visualize_predictions(model, dataloader, device, num_images=6):
    """
    Muestra imágenes del dataset junto a su predicción.
    """
    model.eval()
    images_so_far = 0
    fig = plt.figure(figsize=(10, 10))

    with torch.no_grad():
        for i, (inputs, labels) in enumerate(dataloader):
            inputs = inputs.to(device)
            labels = labels.to(device)

            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)

            for j in range(inputs.size(0)):
                images_so_far += 1
                ax = plt.subplot(num_images//3, 3, images_so_far)
                ax.axis('off')
                ax.set_title(f'Real: {class_names[labels[j]]}\nPred: {class_names[preds[j]]}')
                inp = inputs.cpu().data[j]
                inp = inp.numpy().transpose((1, 2, 0))
                # Desnormalizar
                inp = std * inp + mean
                inp = np.clip(inp, 0, 1)
                plt.imshow(inp)

                if images_so_far == num_images:
                    return

######################################
#   DEFINIR MODELO: MobileNetV2      #
######################################
model_mobilenet = models.mobilenet_v2(weights='IMAGENET1K_V1')  # PyTorch 2.x

# Ajustar la última capa (classifier) a nuestro número de clases
# Por defecto, es una secuencia: classifier = [Dropout(p=0.2, inplace=True), Linear(in_features=1280, out_features=1000, bias=True)]
# Cambiar la capa final a out_features = num_classes (6)
model_mobilenet.classifier[1] = nn.Linear(model_mobilenet.classifier[1].in_features, num_classes)
model_mobilenet = model_mobilenet.to(device)

# Definir función de pérdida y optimizador
criterion = nn.CrossEntropyLoss()
optimizer_mobilenet = optim.Adam(model_mobilenet.parameters(), lr=learning_rate)

######################################
#     ENTRENAR MODELO MOBILE NET     #
######################################
trained_mobilenet, history_mobilenet = train_model(
    model_mobilenet,
    criterion,
    optimizer_mobilenet,
    {'train': dataloaders['train'], 'val': dataloaders['val']},
    device,
    num_epochs=epochs,
    logs_filename='mobilenet_training_log.txt'
)

# Graficar métricas
visualize_metrics(history_mobilenet, title_prefix="MobileNetV2")

######################################
#    EVALUAR EN EL CONJUNTO DE TEST  #
######################################
def val_model(model, dataloader, device):
    model.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs = inputs.to(device)
            labels = labels.to(device)
            outputs = model(inputs)
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
    accuracy = 100 * correct / total
    return accuracy

mobilenet_test_accuracy = test_model(trained_mobilenet, dataloaders['test'], device)
print("Test Accuracy - MobileNetV2:", mobilenet_test_accuracy)

######################################
#  EJEMPLOS DE PREDICCIONES (TETS)   #
######################################
visualize_predictions(trained_mobilenet, dataloaders['test'], device, num_images=6)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import torch

# 0: angry, 1: disgust, 2: fear, 3: happy,
# 4: sad, 5: surprise
emotions_list = ["angry", "disgust", "fear", "happy", "sad", "surprise"]

# mean y std de normalización (ImageNet) para deshacer la normalización al mostrar
mean = [0.485, 0.456, 0.406]
std  = [0.229, 0.224, 0.225]

def visualize_predictions_three_models(models, dataloader, device, num_images=6):
    """
    Muestra imágenes del dataset con la emoción real y la predicción de TRES modelos.
    models: tupla/lista con los 3 modelos (model1, model2, model3)
    dataloader: un DataLoader de prueba o validación
    device: cuda o cpu
    num_images: cuántas imágenes mostrar
    """
    model1, model2, model3 = models

    # Pasar cada modelo a eval
    model1.eval()
    model2.eval()
    model3.eval()

    images_count = 0
    fig = plt.figure(figsize=(12, 14))

    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs = inputs.to(device)
            labels = labels.to(device)

            # Predicciones de cada modelo
            preds1 = model1(inputs).argmax(dim=1)
            preds2 = model2(inputs).argmax(dim=1)
            preds3 = model3(inputs).argmax(dim=1)

            # Para cada imagen del batch
            for j in range(inputs.size(0)):
                images_count += 1
                ax = plt.subplot(num_images, 1, images_count)  # una fila por imagen, para ver las 3 preds
                ax.axis('off')

                # Desnormalizar para mostrar
                inp = inputs[j].cpu().numpy().transpose((1, 2, 0))  # (C,H,W)->(H,W,C)
                inp = (inp * std) + mean
                inp = np.clip(inp, 0, 1)

                # Etiquetas reales y predichas
                real_label = emotions_list[labels[j].item()]
                pred_label_1 = emotions_list[preds1[j].item()]
                pred_label_2 = emotions_list[preds2[j].item()]
                pred_label_3 = emotions_list[preds3[j].item()]

                # Mostrar la imagen y en el título las 3 predicciones
                plt.imshow(inp)
                ax.set_title(
                    f"Real: {real_label}\n"
                    f"ResNet: {pred_label_1},  "
                    f"VGG: {pred_label_2},  "
                    f"MobileNet: {pred_label_3}"
                )

                if images_count == num_images:
                    plt.tight_layout()
                    plt.show()
                    return
```

```python
import torchvision.transforms as transforms

# Transformaciones que coinciden con tu entrenamiento (ej. ImageNet size 224x224)
transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

def predict_emotion_three_models(models, img, device):
    """
    Dada una imagen PIL (img) y una tupla de 3 modelos,
    retorna la emoción predicha por cada modelo.
    """
    # Transformar la imagen
    img_t = transform(img).unsqueeze(0).to(device)  # (1, C, H, W)

    model1, model2, model3 = models
    model1.eval()
    model2.eval()
    model3.eval()

    with torch.no_grad():
        pred1 = model1(img_t).argmax(dim=1).item()
        pred2 = model2(img_t).argmax(dim=1).item()
        pred3 = model3(img_t).argmax(dim=1).item()

    # Emoción textual
    emotion1 = emotions_list[pred1]
    emotion2 = emotions_list[pred2]
    emotion3 = emotions_list[pred3]

    return emotion1, emotion2, emotion3
```

```python
import os
import time
import torch
import torchvision.transforms as transforms
import numpy as np
from PIL import Image

# lista de emociones
emotions_list = ["angry", "disgust", "fear", "happy", "sad", "surprise"]

# Estadísticas de normalización para usar modelos preentrenados en ImageNet
mean = [0.485, 0.456, 0.406]
std  = [0.229, 0.224, 0.225]

# Transformaciones para preparar la imagen
transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor(),
    transforms.Normalize(mean, std)
])

# Ruta donde guardar las predicciones
predictions_path = '/content/drive/MyDrive/IA_ProyectoAgosto/Enero/Predicciones'

###############################################################################
# FUNCIONES PARA GUARDAR LA FOTO EN UNA CARPETA SEGÚN (MODELO + EMOCIÓN)
###############################################################################
def ensure_dir_exists(path):
    """Crea el directorio (y subdirectorios) si no existen."""
    if not os.path.exists(path):
        os.makedirs(path)

def save_prediction_photo(img_pil, emotion_label, model_name, predictions_dir=predictions_path):
    """
    Guarda la imagen (PIL) en la carpeta:
    <predictions_dir>/<model_name>/<emotion_label>/ con un nombre basado en timestamp.
    """
    # Construir la ruta según el modelo y la emoción
    save_path = os.path.join(predictions_dir, model_name, emotion_label)
    ensure_dir_exists(save_path)

    # Generar un nombre de archivo usando timestamp
    timestamp = int(time.time())
    filename = f"prediction_{timestamp}.jpg"
    filepath = os.path.join(save_path, filename)

    # Guardar la imagen
    img_pil.save(filepath)

def predict_emotion(model, img_pil, device):
    """
    Hace la predicción de la emoción usando un modelo (PyTorch).
    Retorna la etiqueta textual de la emoción.
    """
    model.eval()
    # Aplicar las transformaciones y mover a device
    img_t = transform(img_pil).unsqueeze(0).to(device)

    with torch.no_grad():
        outputs = model(img_t)
        _, predicted_idx = torch.max(outputs, dim=1)

    emotion_label = emotions_list[predicted_idx.item()]
    return emotion_label

###############################################################################
#              EJEMPLO DE USO: PREDICCIÓN CON 3 MODELOS Y GUARDADO            #
###############################################################################
def predict_and_save_with_three_models(img_pil, models_tuple, model_names, device):
    """
    Dada una imagen PIL (img_pil), predice la emoción con tres modelos distintos
    y guarda la foto en la carpeta de cada modelo/emoción correspondiente.

    - models_tuple: (modelA, modelB, modelC)
    - model_names:  ["ResNet", "VGG", "MobileNet"] (ejemplo)
    """
    for model, m_name in zip(models_tuple, model_names):
        emotion_label = predict_emotion(model, img_pil, device)
        print(f"Modelo: {m_name}  ->  Emoción predicha: {emotion_label}")
        # Guardar la foto en la carpeta
        save_prediction_photo(img_pil, emotion_label, m_name)

###############################################################################
#       DEMO PARA TOMAR O SUBIR UNA FOTO Y HACER PREDICCIONES CON 3 MODELOS   #
###############################################################################

models_tuple = (trained_resnet, trained_vgg, trained_mobilenet)
model_names = ["ResNet", "VGG", "MobileNet"]

##########################################################
# 1) Tomar foto desde Colab#
##########################################################
from IPython.display import Javascript, display
from google.colab.output import eval_js
import base64
import io

def take_photo(filename='photo.jpg', quality=0.8):
    """
    Toma una foto desde la cámara en Colab y la retorna como PIL.Image.
    """
    js = Javascript('''
      async function takePhoto(quality) {
        const div = document.createElement('div');
        const capture = document.createElement('button');
        capture.textContent = 'Tomar foto';
        div.appendChild(capture);

        const video = document.createElement('video');
        video.style.display = 'block';
        const stream = await navigator.mediaDevices.getUserMedia({video: true});
        document.body.appendChild(div);
        div.appendChild(video);
        video.srcObject = stream;
        await video.play();

        // Ajustar la altura del iframe a la vista de la cámara
        google.colab.output.setIframeHeight(document.documentElement.scrollHeight, true);

        await new Promise((resolve) => capture.onclick = resolve);

        const canvas = document.createElement('canvas');
        canvas.width = video.videoWidth;
        canvas.height = video.videoHeight;
        canvas.getContext('2d').drawImage(video, 0, 0);
        stream.getVideoTracks()[0].stop();
        const dataUrl = canvas.toDataURL('image/jpeg', quality);
        div.remove();
        return dataUrl;
      }
    ''')
    display(js)
    data = eval_js('takePhoto({})'.format(quality))
    if data:
        binary = base64.b64decode(data.split(',')[1])
        with open(filename, 'wb') as f:
            f.write(binary)
        return Image.open(filename)
    else:
        print("No se capturó la foto.")
        return None

# ubir una foto desde tu PC:
from google.colab import files

def upload_photo():
    """
    Sube una foto desde tu computadora y la retorna como PIL.Image.
    """
    uploaded = files.upload()  # se abrirá el navegador para seleccionar
    for fname in uploaded.keys():
        return Image.open(fname)
    return None

# 2) Tomar la foto
img_pil = take_photo()

# 3) O subir la foto
#img_pil = upload_photo()

# 4) Si la foto se obtuvo correctamente, predecir y guardar
if img_pil:
  predict_and_save_with_three_models(img_pil, models_tuple, model_names, device)
```
