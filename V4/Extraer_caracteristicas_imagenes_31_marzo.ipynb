```python
import os
import pandas as pd
from keras.preprocessing import image
from keras.applications.vgg16 import VGG16, preprocess_input
from keras.models import Model
import numpy as np

# Cargar el modelo VGG16 preentrenado
base_model = VGG16(weights='imagenet')
model = Model(inputs=base_model.input, outputs=base_model.get_layer('fc1').output)

def extract_features(img_path):
    img = image.load_img(img_path, target_size=(224, 224))
    x = image.img_to_array(img)
    x = np.expand_dims(x, axis=0)
    x = preprocess_input(x)
    features = model.predict(x)
    return features.flatten()

def load_images_and_extract_features(directory):
    # Diccionario para almacenar los datos
    data = {'ID': [], 'Name': [], 'Number': [], 'Features': []}

    # Cargar todas las imágenes en el directorio
    for filename in os.listdir(directory):
        if filename.endswith(".jpg"):  # Asegúrate de procesar solo archivos .jpg
            file_path = os.path.join(directory, filename)
            features = extract_features(file_path)

            # Extraer identificadores del nombre del archivo
            # Suponiendo formato: photo_NombreCompleto_#.jpg
            parts = filename.split('_')
            if len(parts) < 3:
                continue  # Si el nombre del archivo no cumple el formato esperado, lo ignoramos
            number = parts[-1].split('.')[0]  # Parte numérica después del último '_'
            name = '_'.join(parts[1:-1])  # Combina las partes del nombre que pueden incluir espacios y números

            # Almacenar datos en el diccionario
            data['ID'].append(number)
            data['Name'].append(name)
            data['Number'].append(number)
            data['Features'].append(features.tolist())

    # Convertir el diccionario a DataFrame
    df = pd.DataFrame(data)
    return df

# Ruta al directorio con las imágenes
directory = '/content/drive/My Drive/200Estudiantes/'

# Cargar imágenes, extraer características y guardar en DataFrame
df = load_images_and_extract_features(directory)

# Guardar el DataFrame en un archivo CSV
df.to_csv('/content/drive/My Drive/200Estudiantes/Estudiantes_features.csv', index=False)
```
